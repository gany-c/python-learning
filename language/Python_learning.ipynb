```python
first_set = {1, 2, 3}
first_set.add(1)
print(first_set)
print(first_set.add(4))
first_set.remove(3)
print(first_set)
print(2 in first_set)
```

A frozen set is an immutatble set - once created, you cannot add or remove items

```python
third_set = frozenset({1, 5, 99})
print(third_set)
print(type(third_set))
```

Nested Sets - You can only add a frozen set inside a set

```python
set_outer = set()
set_inner = set()
set_inner.add(2)
# set_outer.add(set_inner) - wil not work, TypeError: unhashable type: 'set'
frozen_inner = frozenset(set_inner)
set_outer.add(frozen_inner)
```

Appending in place vs using the pipe operator

```python
print("This operation has no return type, it is an in-place append", set_inner.add(3))
```

```python
print("This operation returns a new set", set_inner | {4} )
```

## Default Dict

Transform an array of objects into an dict of arrays

```python
transaction_list = [
    {'date': '2023-11-01', 'product': 'A', 'amount': 100, 'customer': 'Alice'},
    {'date': '2023-11-02', 'product': 'B', 'amount': 150, 'customer': 'Bob'},
    {'date': '2023-11-03', 'product': 'A', 'amount': 200, 'customer': 'Charlie'},
    {'date': '2023-11-04', 'product': 'C', 'amount': 50, 'customer': 'David'},
    {'date': '2023-11-05', 'product': 'B', 'amount': 75, 'customer': 'Eve'},
]
```

```python
trans_map = {}
for trans in transaction_list:
    product = trans['product']
    if product not in trans_map:
        trans_map.update({product: []})
    trans_map[product].append(trans)
print(trans_map)    
```

Do this better with a defaultdict

```python
from collections import defaultdict
trans_map = defaultdict(list)
for trans in transaction_list:
    product = trans['product']
    trans_map[product].append(trans)
print(trans_map)
```

# Global Interpreter Lock - GIL

1. GIL is the Global Interpreter Lock in Python, it is the default behavior and cannot be overridden.
2. GIL makes sure that only one thread can access the Python Interpreter at a time.
3. As an analogy in Java, even if you have code that is multi threaded, if the machine has a single core, only one thread runs at a time. In Python, the GIL makes even a multi-core machine behave like a single core machine.
4. So, multi threading in Python is useful only if the threads do remote or I/O operations.
5. Alternately, you can do multi-processing. Then each process will have its own Interpreter and GIL
6. There is talk about diminishing or eliminating the GIL. Starting with Python 3.13, CPython introduced an experimental free-threaded build where the GIL can be disabled.

### Keyword Args or Kwargs

```python
# There is nothing particularly different in the invocation, not even the default values, it is just the invocation
def fourth_function(param1, param2):
    print(f"param1 = {param1}, param2 = {param2}")
    
```

```python
# fourth_function(param2="hello", 2) Does not work - positional argument follows keyword argument
fourth_function(3, param2="hello")
```

```python
fourth_function(param2="Hello", param1="Hi") # works, the order is maintained despite being jumbled in invocation
```

```python
def fifth_function(*params, **key_params):
    print("Positional params = ", params)
    print("Keyword params = ", key_params)
```

```python
fifth_function(1, 2, 3)
```

```python
fifth_function(1, 2, gauge=12, barrel=2)
```

# Typing Hints

With Python 3.9, it looks there are 2 ways to provide type hints

https://stackoverflow.com/questions/39458193/using-list-tuple-etc-from-typing-vs-directly-referring-type-as-list-tuple-etc 

```python
# using the lower case list as the type hint
def square_list(in_list: list[int]) -> list[int]:
    return [x*x for x in in_list]
```

```python
print(square_list([3, 5, 7]))
```

```python
from typing import List
def cube_list(in_list: list[int]) -> list[int]:
    return [x*x*x for x in in_list]
    
```

```python
print(cube_list([2, 3, 5]))
```

Type aliases are also available
https://docs.python.org/3/library/typing.html#type-aliases

you can use the "type" keyword to create a new type; but it is a type alias—not a genuinely new type

```python
Vector = List[float]
```

```python
def scale(scalar: float, vector: Vector) -> Vector:
    return [scalar * element for element in vector]
```

```python
scaled_vector = scale(2.0, [1, -4.2, 5.4])
print(scaled_vector)
```
