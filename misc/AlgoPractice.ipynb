```python
def merge_sorted(a, b):
    if not a or not b:
        return None

    output_len = len(a) + len(b)
    output = [None] * output_len

    output_ptr = 0
    a_ptr = 0
    b_ptr = 0

    while a_ptr < len(a) and b_ptr < len(b):

        if a[a_ptr] <= b[b_ptr]:
            output[output_ptr] = a[a_ptr]
            a_ptr = a_ptr + 1
        else:
            output[output_ptr] = b[b_ptr]
            b_ptr = b_ptr + 1

        output_ptr = output_ptr + 1

    if a_ptr < len(a):
        while a_ptr < len(a):
            output[output_ptr] = a[a_ptr]
            a_ptr = a_ptr + 1
            output_ptr = output_ptr + 1
    elif b_ptr < len(b):
        while b_ptr < len(b):
            output[output_ptr] = b[b_ptr]
            b_ptr = b_ptr + 1
            output_ptr = output_ptr + 1        
            

    
    return output
```

```python
"""
Given 
   a Text: "aaba"  
   And a vocab dict: {"a": 1, "aa": 2, "ab": 3}
 Write a function that returns a valid tokenization of the text. E.g. {1,3,1}
 The given text in it's entirety is always tokenizable

"""

def find_valid_prefixes(Text, vocab_dict):
    """
    time complexity would be the length of the vocab items
    O(keys)
    """
    prefixes_list = []
    for key, value in vocab_dict.items():
        if Text.startswith(key):
            prefixes_list.append(key)
    return prefixes_list        


"""
aaaaaa {"a": 1, "aa": 2, "ab": 3} 
Worst case time complexity - (number of keys in dict)^2 * number of tokens in the string
"aaaa.....abaaa", vocab : {"b": 1, "aa": 2, "aaa": 3}
"""
def encode_to_tokens(Text, vocab_dict):

    prefixes_list = find_valid_prefixes(Text, vocab_dict)
    if prefixes_list is empty: 
            return None
    else:
        for prefix in prefixes_list: 
            suffix = Text.removeprefix(prefix)
            if len(suffix) == 0:
                # break out at the first valid answer
                # return as an array
                return [vocab_dict[prefix]]
            else:    
                output_tokens = encode_to_tokens(suffix, dict)
                if output_tokens is not None: 
                    return  output_tokens.insert(0, vocab_dict[prefix])
                else 
                    # continuing with the loop, because there may be other prefixes that worked
                    continue
        return None     
```

Takeaways -

1. You missed one of the exit conditions - the prefix exactly matches the full string
2. You struggled with giving the time complexity when there recursion inside an if condition inside a loop. Solution May be give the worst case complexity, i.e. given any input calculate the maximum possible time something can run, ignore the logical conditions.
